````
#Pregunta 1
print("Sobrevivientes=",df.survived.sum())
print('Muertos=',df.shape[0]-df.survived.sum())
print("Total pasajeros=",df.shape[0])
````

````
#Pregunta 2: Cuántos pasajeros de primera, segunda y tercera clase sobrevienieron. Calcular el porcentaje respecto a su clase.
#Sobrevievientes por clase
s1c = df[np.logical_and(df.pclass == 1 ,df.survived ==1)]
s2c = df[np.logical_and(df.pclass == 2 ,df.survived ==1)]
s3c = df[np.logical_and(df.pclass == 3 ,df.survived ==1)]

print(s1c.shape[0],s2c.shape[0],s3c.shape[0])
#porcentajes
p1c = (s1c.shape[0]/df[df.pclass == 1].shape[0])*100
p2c = (s2c.shape[0]/df[df.pclass == 2].shape[0])*100
p3c = (s3c.shape[0]/df[df.pclass == 3].shape[0])*100
print("{:.1f}".format(p1c),"%","{:.1f}".format(p2c),"%","{:.1f}".format(p3c),"%")
````
